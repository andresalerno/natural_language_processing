## TC.1 — Frase, documento, corpus e corpora em diferentes contextos

### Definições adotadas

| Elemento | Definição |
|---|---|
| **Frase (sentença)** | Menor unidade textual que expressa uma ideia relativamente completa; sequência de palavras normalmente delimitada por pontuação forte (`.`, `?`, `!`) ou por quebra de turno. É a unidade típica de tarefas como *tokenization*, *parsing*, tradução e análise de sentimento em nível de sentença. |
| **Documento** | Unidade de texto coesa e autocontida, produzida em **um único contexto comunicativo** (um autor, um evento, uma interação). É formado por uma ou mais frases e costuma carregar **metadados próprios** (autor, data, título, categoria, avaliação…). |
| **Corpus** | Coleção **estruturada e finita** de documentos, reunidos segundo um **critério explícito** (gênero, época, fonte, língua, domínio). Em geral acompanha metadados e, às vezes, anotações linguísticas (POS, árvores sintáticas). Serve de base para estudo linguístico ou para treino/avaliação de modelos de PLN. |
| **Corpora** | Plural de *corpus*: **conjunto de dois ou mais corpora** tratados em conjunto — frequentemente de fontes, gêneros ou línguas diferentes — para estudos comparativos ou para ampliar a cobertura de um modelo. |

> Relação hierárquica: **palavra ⊂ frase ⊂ documento ⊂ corpus ⊂ corpora**.

## PC.1 — Atributos dos corpora `machado`, `mac_morpho` e `floresta` (NLTK)

O programa abaixo:

1. baixa os três corpora com `nltk.download`;
2. exibe os **atributos comuns** de cada *CorpusReader* (classe, diretório raiz, codificação, número de arquivos, `fileids`, métodos de acesso disponíveis);
3. exibe **atributos específicos** de cada um (categorias, contagem de palavras/frases/parágrafos, conjunto de etiquetas, `README` e amostras de conteúdo);
4. é seguido de uma célula de texto descrevendo **origem, metadados, anos dos textos e ano de produção** de cada corpus.

> **Nota técnica:** na versão 3.9.1 do NLTK o *loader* `nltk.corpus.machado` está com um *bug* (o `PortugueseCategorizedPlaintextCorpusReader` não inicializa a superclasse, gerando `AttributeError: _fileids`). Por isso o corpus Machado é reconstruído manualmente com `CategorizedPlaintextCorpusReader` a partir dos mesmos arquivos baixados — o resultado é idêntico ao do *reader* oficial.


```python
import nltk
from nltk.corpus.reader import CategorizedPlaintextCorpusReader

# 1) Garantir os corpora no disco ------------------------------------------------
# IMPORTANTE: no NLTK 3.9.1, chamar nltk.download() para um corpus que JA foi
# carregado em memoria corrompe o ponteiro interno do reader (_root perde o
# atributo _path) e qualquer acesso a `corpus.root` passa a lancar
#   AttributeError: 'FileSystemPathPointer' object has no attribute '_path'
# Por isso so baixamos o que ainda nao existe -- assim, ao reexecutar as
# celulas, nltk.download() nunca e chamado de novo.
# (Se o kernel ja estiver nesse estado, reinicie-o e rode tudo de novo.)
def garantir(pacote, recurso):
    try:
        nltk.data.find(recurso)
    except LookupError:
        nltk.download(pacote, quiet=True)

garantir("machado", "corpora/machado")
garantir("mac_morpho", "corpora/mac_morpho")
garantir("floresta", "corpora/floresta")

from nltk.corpus import mac_morpho, floresta   # noqa: E402  (apos garantir o download)


def _raiz(corpus):
    """str do diretorio raiz do corpus, resistente ao bug do _path no NLTK 3.9.1."""
    try:
        return str(corpus.root)
    except AttributeError:
        return "(indisponivel - reinicie o kernel e rode tudo de novo)"


def resumo_generico(nome, corpus):
    """Exibe, de forma compacta, os atributos comuns a qualquer CorpusReader."""
    fids = corpus.fileids()        # dispara o carregamento "preguiçoso" do reader
    try:
        enc = corpus.encoding(fids[0])
    except Exception:
        enc = "n/d"
    metodos = [m for m in ("raw", "words", "sents", "paras", "tagged_words",
                           "tagged_sents", "parsed_sents", "categories")
               if hasattr(corpus, m)]
    print(f"### {nome}  (reader: {type(corpus).__name__})")
    print("raiz     :", _raiz(corpus))
    print("cod./arq.:", enc, "/", len(fids), "arquivo(s)  ->", fids[:4],
          "..." if len(fids) > 4 else "")
    print("metodos  :", metodos)
```

```python
# =========================================================================
# (a) MACHADO  --  Obra Completa de Machado de Assis
# =========================================================================
# Reconstrucao manual do reader (contorna o bug do loader oficial no NLTK 3.9.1)
raiz_machado = nltk.data.find("corpora/machado.zip/machado/")
machado = CategorizedPlaintextCorpusReader(
    raiz_machado, r"(?!\.).*\.txt",
    cat_pattern=r"([a-z]+)/.*",    # categoria = subdiretorio (romance, contos, ...)
    encoding="latin-1",
)

resumo_generico("(a) MACHADO", machado)

# --- Atributos especificos ----------------------------------------------------
print("\ncategorias (generos):")
for cat in machado.categories():
    fids = machado.fileids(categories=cat)
    print(f"  {cat:11s} {len(fids):3d} arq.  ex.: {fids[0]}")
print("total de palavras no corpus:",
      sum(len(machado.words(f)) for f in machado.fileids()))

# README: catalogo com titulo + ANO de publicacao (secao dos romances)
readme_machado = nltk.data.find(
    "corpora/machado.zip/machado/README").open().read().decode("latin-1")
romances = readme_machado.split("Romance", 1)[1].split("Poesia", 1)[0].strip()
print("\n----- README: romances (titulo + ano) -----")
print(romances)

# Um DOCUMENTO do corpus: o romance "Dom Casmurro" (1899)
doc = "romance/marm08.txt"
print(f"\n----- Documento de exemplo: {doc} -----")
print(f"paragrafos={len(machado.paras(doc))}  frases={len(machado.sents(doc))}"
      f"  palavras={len(machado.words(doc))}")
print("metadados (inicio do arquivo):", " ".join(machado.words(doc)[:6]))
print("frase de exemplo            :", " ".join(machado.sents(doc)[6]))
```

### (a) Machado — características

| Aspecto | Descrição |
|---|---|
| **O que é** | *Obra Completa* de Machado de Assis (1839–1908), o principal escritor do Realismo brasileiro. É um corpus **literário monolíngue (português do Brasil), de autor único**. |
| **Origem do texto** | Digitalização das *Obras Completas de Machado de Assis*, vol. I–III, editora Nova Aguilar (Rio de Janeiro, 1994), tendo como base as edições originais da Editora Garnier. Disponibilizada em **domínio público** pelo portal do Ministério da Educação (`machado.mec.gov.br`). |
| **Estrutura / metadados** | 246 arquivos `.txt` organizados em 8 subdiretórios, que funcionam como **categorias / gêneros**: `romance`, `conto`s (`contos`), `cronica`, `critica`, `poesia`, `teatro`, `traducao` e `miscelanea`. O arquivo `README` (e `CONTENTS`) traz o **catálogo completo com título e ano de cada obra**; as primeiras linhas de cada `.txt` repetem gênero, título e ano. |
| **Anos dos textos** | Obras publicadas entre **1854 e 1939** (póstumas/dispersas). Exemplos: *Crisálidas* (1864), *Ressurreição* (1872), *Memórias Póstumas de Brás Cubas* (1881), *Quincas Borba* (1891), *Dom Casmurro* (1899), *Esaú e Jacó* (1904), *Memorial de Aires* (1908). |
| **Anotação** | Nenhuma — é **texto puro** (*plaintext*). O NLTK oferece acesso a `raw`, `words`, `sents`, `paras` e às `categories`. |
| **Ano de produção do corpus** | Compilação eletrônica feita pelo NILC/USP a partir da edição de 1994; incorporada ao NLTK Data (pacote `machado`, ~5 MB). |
| **Codificação / licença** | ISO-8859-1 (`latin-1`); domínio público. |
| **Uso típico** | Estudos de estilo e autoria, modelagem de linguagem histórica do português, geração de texto, análise literária computacional. |


```python
# =========================================================================
# (b) MAC-MORPHO  --  noticias em portugues do Brasil anotadas com POS
# =========================================================================
resumo_generico("(b) MAC-MORPHO", mac_morpho)

# O README esta gravado em UTF-8, mas mac_morpho.readme() o decodifica com a
# codificacao do corpus (latin-1) e gera acentos quebrados ("NÃºcleo...").
# Lemos o arquivo manualmente em UTF-8.
readme_mm = nltk.data.find("corpora/mac_morpho/README").open().read().decode("utf-8")
print("\n----- README (lido em UTF-8) -----")
print(readme_mm.strip())

# --- Atributos especificos ----------------------------------------------------
etiquetas = sorted({tag for _, tag in mac_morpho.tagged_words()})
gramaticais = [t for t in etiquetas if t[:1].isalpha()]
print(f"\npalavras={len(mac_morpho.words())}  frases={len(mac_morpho.tagged_sents())}"
      f"  etiquetas_distintas={len(etiquetas)}")
print("etiquetas gramaticais (amostra):", gramaticais[:22])
print(f"\nfrase etiquetada de exemplo (fileid = {mac_morpho.fileids()[0]}):")
print(mac_morpho.tagged_sents()[0][:12])
```

### (b) Mac-Morpho — características

| Aspecto | Descrição |
|---|---|
| **O que é** | Corpus **jornalístico do português do Brasil anotado morfossintaticamente** (*part-of-speech tagging*). É a base de referência para treino/avaliação de etiquetadores POS em PT-BR. |
| **Origem do texto** | ~1.167.183 palavras de textos jornalísticos extraídos de **dez seções do jornal *Folha de S.Paulo*, ano de 1994** (caderno de agropecuária, economia, esporte etc.). |
| **Estrutura / metadados** | Na distribuição do NLTK: **109 arquivos** `.txt` (versão para treino de *taggers*), ~1,17 milhão de tokens e ~51,4 mil frases. O **nome de cada arquivo codifica a seção e a data** do texto (ex.: `ag94ab12.txt` ≈ seção "agropecuária", abril de 1994). |
| **Anotação** | Uma etiqueta morfossintática por *token*. Conjunto de ~22 etiquetas básicas — `N` (substantivo), `V` (verbo), `ART`, `PREP`, `ADJ`, `ADV`, `NPROP` (nome próprio), `NUM`, `KC`/`KS` (conjunções), `PROPESS`, `PROADJ`, `PCP`, `CUR` (moeda) … — acrescidas de sufixos de contração/estrutura (`|+`, `|EST`, `|[`, `|]`), o que eleva o total a ~80 rótulos distintos no arquivo. Acesso via `words`, `sents`, `tagged_words`, `tagged_sents`. |
| **Anos dos textos** | 1994. |
| **Produção / ano de produção** | Produzido pelo **NILC – Núcleo Interinstitucional de Linguística Computacional** (USP São Carlos), com UFSCar e UNESP/Araraquara, no âmbito do projeto **Lácio-Web** (início dos anos 2000). Distribuído com permissão do NILC. |
| **Codificação / licença** | `latin-1`; uso acadêmico, distribuído com permissão. |
| **Uso típico** | Treino e avaliação de POS *taggers* para o português; estudos de frequência de classes gramaticais. |


```python
# =========================================================================
# (c) FLORESTA SINTACTICA  --  treebank do portugues (PT europeu + PT-BR)
# =========================================================================
import io
import contextlib

resumo_generico("(c) FLORESTA SINTACTICA", floresta)

print("\n----- README (cabecalho) -----")
print(floresta.readme()[:210].rstrip())
print("   [...]  chave das etiquetas: http://visl.sdu.dk/visl/pt/portsymbol.html")

# Percorrer TODA a Floresta imprime mensagens "Bad tree detected..." em
# sys.stderr: algumas frases do .ptb tem parenteses desbalanceados e o
# BracketParseCorpusReader as substitui por uma analise sintatica "plana".
# Nao e erro do nosso codigo; capturamos as mensagens so para manter a saida
# limpa e contamos quantas frases foram recuperadas.
_avisos = io.StringIO()
with contextlib.redirect_stderr(_avisos):
    n_palavras = len(floresta.words())
    frases = floresta.parsed_sents()
    n_frases = len(frases)
    etiquetas = sorted({tag for _, tag in floresta.tagged_words()})
    arvore = frases[0]
n_recup = _avisos.getvalue().count("Bad tree detected")

# --- Atributos especificos ----------------------------------------------------
print(f"\npalavras={n_palavras}  frases={n_frases}  etiquetas_distintas={len(etiquetas)}")
print(f"frases recuperadas com parse 'plano': {n_recup} ({n_recup / n_frases:.1%})")
print("etiquetas 'funcao+categoria' (amostra):",
      [t for t in etiquetas if any(c.isalpha() for c in t)][:8])
print("\nfrase 0 - tagged:", floresta.tagged_sents()[0])
print("frase 0 - arvore:", arvore)
arvore.pretty_print()
```

### (c) Floresta Sintáctica — características

| Aspecto | Descrição |
|---|---|
| **O que é** | *Treebank* (banco de árvores sintáticas) do português — o subconjunto distribuído no NLTK corresponde ao **"Bosque"**, a parte revista manualmente do projeto, na **versão 7.4**, em **formato Penn Treebank**. |
| **Origem do texto** | **Projecto Floresta Sintá(c)tica**, da [Linguateca](https://www.linguateca.pt/Floresta/), em parceria com o grupo VISL (Universidade do Sul da Dinamarca). | **Estrutura / metadados** | No NLTK: **1 arquivo** (`Floresta_7.4.ptb`), ~211,8 mil palavras e **~9.266 frases com análise sintática completa**. O `README` traz a **chave das etiquetas** (padrão VISL, `http://visl.sdu.dk/visl/pt/portsymbol.html`). |
| **Anotação** | Análise sintática **em constituintes**, combinando, em cada nó/*token*, **função sintática + categoria morfológica**. As etiquetas de *token* têm a forma `função+categoria` — ex.: `>N+art` (artigo, modificador pré-nominal), `H+n` (núcleo nominal), `N<+adj` (adjetivo pós-nominal), `P+v-fin` (verbo finito). Há centenas de rótulos distintos (`SUBJ`, `ACC`, `PIV`, `ADVL`, `APP`, `UTT`, `NPHR`…). Acesso via `words`, `sents`, `tagged_words`, `tagged_sents` e **`parsed_sents`** (objetos `nltk.Tree`). |
| **Anos dos textos** | Notícias do final da década de 1990. |
| **Produção / ano de produção** | Projeto iniciado em **2000**; versão 7.4 distribuída com permissão. Mantido pela Linguateca (Portugal). |
| **Codificação / licença** | `ISO-8859-15` (latin-9); distribuído com permissão para uso acadêmico. |
| **Uso típico** | Treino e avaliação de *parsers* e *chunkers*, extração de gramáticas, estudos de sintaxe do português. |
